```python
import pandas as pd
import plotly.graph_objects as go

mines = pd.read_csv("./data_combined/Producing Mines.csv")
canada = pd.read_csv("./data_combined/canada_raster_stack (2).csv")
```

```python
mines = mines[["province", "longitude", "latitude", "Commodity"]]
mines["province"].value_counts()
```

```python
print(canada.shape)
print(canada.columns)
print(canada.isna().mean())
canada.head()
```

```python
canada.drop(columns=['pixel_id', 'col_idx', 'row_idx', 'x_3978', 'y_3978'], inplace=True, errors='ignore')
```

```python
fig = go.Figure()

fig.add_trace(go.Scattermapbox(
    lat=canada["wgs84_lat"],
    lon=canada["wgs84_long"],
    mode="markers",
    opacity=0.2,
    name="canada raster stack data",
))

fig.add_trace(go.Scattermapbox(
    lat=mines["latitude"],
    lon=mines["longitude"],
    mode="markers",
    opacity=0.5,
    name="Mines data",
))

fig.update_layout(
    mapbox=dict(
        style="carto-positron",
        zoom=2.5,
        center=dict(lat=60, lon=-100)
    ),
    margin=dict(l=0, r=0, t=0, b=0)
)
fig.show()
```

```python
mines_pivoted = (
    mines
    .assign(value=True)
    .pivot_table(
        index=["province", "longitude", "latitude"],
        columns="Commodity",
        values="value",
        fill_value=False,
        aggfunc="any"
    )
    .reset_index()
)
mines_pivoted = mines_pivoted.astype({col: "int8" for col in mines_pivoted.columns if col not in ["province", "longitude", "latitude"]})
mines_pivoted.drop(columns=['examine', 'gq.mines.gouv.qc.ca', 'https:', 'lime', 'documents', 'Aggregate'], inplace=True, errors='ignore')
mines_pivoted
```

```python
mines_pivoted.to_csv("./data_combined/mines_pivoted.csv", index=False)
```

```python
import geopandas as gpd

mines_pivoted_gdf = gpd.GeoDataFrame(
    mines_pivoted,
    geometry=gpd.points_from_xy(mines_pivoted["longitude"], mines_pivoted["latitude"]),
    crs="EPSG:4326" 
)

canada_gdf = gpd.GeoDataFrame(
    canada,
    geometry=gpd.points_from_xy(canada["wgs84_long"], canada["wgs84_lat"]),
    crs="EPSG:4326"
)

mines_pivoted_gdf = mines_pivoted_gdf.to_crs("EPSG:3978") # distance in meters
canada_gdf = canada_gdf.to_crs("EPSG:3978")

result = mines_pivoted_gdf.sjoin_nearest(canada_gdf, distance_col="distance")
result.drop(columns=['wgs84_long', 'wgs84_lat', 'geometry', 'index_right'], inplace=True, errors="ignore")
result
```

```python
# indices of geographic data points that matched to a mine
used_geo_idx = result["index_right"].unique() 

# rows that did not match to a mine within 10km
negative_samples = canada_gdf.loc[~canada_gdf.index.isin(used_geo_idx)].copy()
negative_samples.drop(columns=['geometry'], inplace=True, errors="ignore")
negative_samples.rename(columns={"wgs84_long": "longitude", "wgs84_lat": "latitude"}, inplace=True)
negative_samples.to_csv("./data_combined/unlabeled_samples.csv", index=False)
```

```python
result_filtered = result[result["distance"] <= 10_000].copy()
result_filtered["distance"].describe()
```

```python
df = pd.read_csv("./data_combined/geo_and_mineral_data.csv")
df.head()
```

```python
import requests

def get_elevation(lat, lon):
    url = "http://geogratis.gc.ca/services/elevation/cdem/altitude"
    params = {"lat": lat, "lon": lon}
    
    try:
        r = requests.get(url, params=params, timeout=5)
        r.raise_for_status()
        return r.json().get("altitude")
    except Exception:
        return None
    
from time import sleep
from tqdm import tqdm

tqdm.pandas()

def safe_get(row):
    sleep(0.05) 
    return get_elevation(row["latitude"], row["longitude"])

df["elevation"] = df.progress_apply(safe_get, axis=1) # type: ignore
```

```python
df.to_csv("./data_combined/geo_and_mineral_data.csv", index=False)
```
